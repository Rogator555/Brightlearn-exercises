```python
import pandas as pd
```

```python
df = pd.read_csv('retail_sales_dataset.csv')
df.head()
```

### Product category with highest revenue:

```python
df.sort_values("Total Amount", ascending = False).head()
```

### Product category with lowest revenue:

```python
df.sort_values("Total Amount", ascending = True).head()
```

The answer is clothing for both highest revenue and lowest revenue

### Which gender spends more on average?

```python
df.groupby("Gender")["Total Amount"].mean()
```

The answer is Females spend more on average

### Show top 5 transactions

### Assuming the Total amount excludes VAT, create:

### VAT column 

### Final Amount column 

### Average Transaction value 

### Most common product category

```python
# VAT Column

df["VAT"] = df["Total Amount"] * 0.15
df.head(3)
```

```python
#Final amount column

df["Final Amount"] = df["Total Amount"] + df["VAT"]
df.head(3)
```

```python
#Average Tansaction value

print(df["Final Amount"].mean())
```

```python
#Most common product category

df["Product Category"].mode()
```
