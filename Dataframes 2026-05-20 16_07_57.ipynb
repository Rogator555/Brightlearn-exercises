```python
#Creating a data using dictionary

dict_data = {"Name": ["Lerato", "Peggy", "Ane", "Rogator"],
             "Age" : [32, 28, 15, 33],
             "City": ["Joburg", "Cape Town", "Durban","PE"]
             }
```

Now change the above data into dataframe using pandas

```python
import pandas as pd  #required to convert dict into dataframe

df = pd.DataFrame(dict_data)
print(df)
```

```python
# How to access specific columns, instead of the entire dataframe

df["Product Category"]
```

```python
df[["Product Category", "Gender"]]
```

```python
df[["Product Category", "Age", "Gender"]].head()
```

```python
#Another way would be to access the column then place an object in front as a variable.

new_columns = df[["Age","Gender","Customer ID"]]
print(new_columns.head())
```

```python
#We can also access the rows only using the iloc() fucntion.

df.iloc[0]  #to access the first row and values
```

```python
#Access the last row

df.iloc[-1].head()
```

```python
# How to access multiple rows, the beauty of using dataframes is we don't need to loop like lists!

df.iloc[0:4].head() #gives the first four rows
```

```python
# Access column less than age 20

df["Age"].head() < 20  #returns a boolean because it's a condition.
```

```python
df[df["Age"] < 20]
```

### Now calling rows by age, in this case condition is age is less than 20

```python
#Run the same code and reduce rows to 5 only

df[df["Age"] <20].head()
```

### NB!!! iloc accessess rows and loc accesses columns using index position

```python
#Group or count how many transactions are in each product category using value_counts()function

df["Product Category"].value_counts()
```

```python
# Now group each a specific column by another one using groupby() function

df.groupby("Product Category")["Age"].max()
```

```python
df.groupby("Product Category")["Age"].mean()
```
